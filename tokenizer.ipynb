```python
text = "Ｕｎｉｃｏｄｅ! 🅤🅝🅘🅒🅞🅓🅔‽ 🇺‌🇳‌🇮‌🇨‌🇴‌🇩‌🇪! 😄 The very name strikes fear and awe into the hearts of programmers worldwide. We all know we ought to “support Unicode” in our software (whatever that means—like using wchar_t for all the strings, right?). But Unicode can be abstruse, and diving into the thousand-page Unicode Standard plus its dozens of supplementary annexes, reports, and notes can be more than a little intimidating. I don’t blame programmers for still finding the whole thing mysterious, even 30 years after Unicode’s inception."
tokens = text.encode("utf-8")
tokens = [*map(int, tokens)]

print("=== TEXT ===")
print(text)
print(len(text))
print("=== TOKENS ===")
print(tokens)
print(len(tokens))
```

```python
"""
Find pairs of tokens that appear together in the text
"""
from collections import Counter

pairs = list(zip(tokens, tokens[1:]))
counter = Counter(pairs)
print("=== PAIRS ===")
print(sorted(counter.items(), key=lambda x: x[1], reverse=True))
print(len(counter))
```

```python
top_pair = max(counter, key=counter.get)
print("=== TOP PAIR ===")
print(top_pair)
```

```python
def merge(ids, pair, idx):
    new_id = []
    i = 0
    while i < len(ids):
        if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
            new_id.append(idx)
            i += 2
        else:
            new_id.append(ids[i])
            i += 1
    return new_id
merged = merge(tokens, top_pair, 256)
print("=== MERGED ===")
print(merged)
print(len(merged))
```

```python
def get_pairs(ids):
    return Counter(list(zip(ids, ids[1:])))
```

```python
VOCAB_SIZE = 276
num_merges = VOCAB_SIZE - 256

ids = list(tokens)
merges = {}
for i in range(num_merges):
    pairs = get_pairs(ids)
    most_common = max(pairs, key=pairs.get)
    ids = merge(ids, most_common, 256 + i)
    merges[most_common] = 256 + i
print(len(tokens))
print(len(ids))
print(f"compression ratio: {len(tokens) / len(ids):.2f}")
```

```python
vocab = {idx: bytes([idx]) for idx in range(256)}
for (p0, p1), idx in merges.items(): # in 3.7+ items inserted in order of insertion
    vocab[idx] = vocab[p0] + vocab[p1]
```

```python
def decode(ids):
    tokens = b"".join(vocab[idx] for idx in ids)
    return tokens.decode("utf-8", errors="replace")
```

```python
def encode(text):
    tokens = list(text.encode("utf-8"))
    while len(tokens) >= 2:
        pairs = get_pairs(tokens)
        pair = min(pairs, key=lambda p: merges.get(p, float("inf")))
        if pair not in merges:
            break
        idx = merges[pair]
        tokens = merge(tokens, pair, idx)
    return tokens
```

```python
print(decode(encode("hello")))
```

```python
text2 = decode(encode(text))
print(text2 == text)
```
